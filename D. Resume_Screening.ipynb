```python
import PyPDF2

def extract_text_from_pdf(file):
    text = ""
    pdf = PyPDF2.PdfReader(file)
    
    for page in pdf.pages:
        text += page.extract_text()
        
    return text
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r'\W', ' ', text)  # remove special characters
    text = re.sub(r'\s+', ' ', text) # remove extra spaces
    return text
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

skills_list = ["python", "machine learning", "data analysis", "deep learning", "nlp"]

def extract_skills(text):
    doc = nlp(text)
    found_skills = []
    
    for token in doc:
        if token.text in skills_list:
            found_skills.append(token.text)
            
    return list(set(found_skills))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

def vectorize(resume, job_desc):
    tfidf = TfidfVectorizer()
    vectors = tfidf.fit_transform([resume, job_desc])
    return vectors
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def get_similarity(vectors):
    score = cosine_similarity(vectors[0:1], vectors[1:2])
    return score[0][0]
```

```python
from transformers import BertTokenizer, BertModel
import torch

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')

def bert_similarity(text1, text2):
    inputs = tokenizer([text1, text2], return_tensors='pt', padding=True, truncation=True)
    
    outputs = model(**inputs)
    embeddings = outputs.last_hidden_state.mean(dim=1)
    
    similarity = cosine_similarity(
        embeddings[0].detach().numpy().reshape(1, -1),
        embeddings[1].detach().numpy().reshape(1, -1)
    )
    
    return similarity[0][0]
```

```python
def skill_gap(resume_skills, job_skills):
    missing = list(set(job_skills) - set(resume_skills))
    return missing
```

```python
def rank_candidates(resumes, job_desc):
    scores = []
    
    for resume in resumes:
        vectors = vectorize(resume, job_desc)
        score = get_similarity(vectors)
        scores.append(score)
        
    return sorted(scores, reverse=True)
```

```python
import streamlit as st

st.title("AI Resume Intelligence System")

uploaded_file = st.file_uploader("Upload Resume")

job_desc = st.text_area("Enter Job Description")

if uploaded_file and job_desc:
    resume_text = extract_text_from_pdf(uploaded_file)
    cleaned = clean_text(resume_text)
    
    vectors = vectorize(cleaned, job_desc)
    score = get_similarity(vectors)
    
    st.write(f"Match Score: {score*100:.2f}%")
```
